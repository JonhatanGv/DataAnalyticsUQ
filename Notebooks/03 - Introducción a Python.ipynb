Tipos de variables en python

1. Tipo string

```python
nombre = 'Jonhatan'
print (nombre)
print(type(nombre))
```

2. Variable tipo número entero

```python
edad = 18
print(edad)
print(type(edad))
```

3. Variable tipo número decimal

```python
estatura = 1.8
print(estatura)
print(type(estatura))
```

4. Variable tipo booleano

```python
es_mayor_de_edad = True
es_menor_de_edad = False
print(es_mayor_de_edad)
print(es_menor_de_edad)
print(type(es_mayor_de_edad))
```

Operaciones básicas matemáticas en python

```python
#Suma
print(2+2)
#Resta
print(2-2)
#Multiplicacion
print(2*2)
#Division
print(2/2)
#Potencia
print(2**2)
#Division entera
print(2//2)
#Modulo
print(2%2)
```
